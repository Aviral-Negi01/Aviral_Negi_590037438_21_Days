```python
fruits = ['apple', 'banana', 'mango'] # fruits is a list containing 3 items
print(fruits)
     
```

```python

fruits = ['apple', 'banana']
fruits.append('mango') #.append() adds an item to the end of a list.
print(fruits)
```

```python
numbers = [5, 2, 9, 1]
numbers.sort()#.sort() arranges numbers from smallest to largest.
print(numbers)
```

```python
items = ['pen', 'book', 'bag']
print(items[1])
print(items[:2]) # starting from index 0 to not including 2
     
```

```python
coordinates = (12.9, 77.6) #tuple uses parentheses
print(coordinates[0], coordinates[1])
```

```python
unique_numbers = set([1, 2, 2, 3, 3])# set stores unique values and removes duplicates
print(unique_numbers)
```

```python
a = {1, 2, 3}
b = {3, 4, 5}
print(a | b) # | means union 
print(a & b) # & means intersection
```

```python
shopping = ['rice', 'tea', 'soap']
print('tea' in shopping) # 'in' checks whether an item exists inside a collection
```

```python
numbers = [1, 2, 3, 4]
print(numbers + [5, 6]) # + operator combines two lists
```

```python
values = [10, 20, 30]
values[1] = 25 # since index 1 contains 20 it replaces 20 with 25
print(values)
```
