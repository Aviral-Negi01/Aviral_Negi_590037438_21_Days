```python
count = 1
while count <= 10:
    print(count)
    count += 1
```

```python
number = 10
while number > 0:
    print(number)
    number -= 2 # number = number - 2
```

```python
total = 0
number = 1
while number <= 5:
    total += number # total = total + number
    number += 1 # number = number + 1
print(total)
```

```python
text = 'Python'
index = 0
while index < len(text):
    print(text[index])
    index += 1
```

```python
number = 1
while number < 20:
    number *= 2
print(number)
```

```python
items = ['pen', 'book', 'bag']
while items:
    print(items.pop())
```

```python
attempts = 0
while attempts < 3:
    print('Attempt', attempts + 1)
    attempts += 1
```

```python
number = 1
while True:
    if number == 4:
        break
    print(number)
    number += 1
```

```python
number = 0
while number < 5:
    number += 1
    if number == 3:
        continue
    print(number)
```

```python
balance = 100
days = 0
while balance < 150:
    balance += 10
    days += 1
print(days, balance)
```
