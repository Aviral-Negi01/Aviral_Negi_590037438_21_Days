# 21 Days Python Challenge
## Day 9 🤩

```python
def greet():
    print('Hello!')
greet()
```

```python
def greet_person(name):
    print(f'Hello, {name}!')
greet_person('Asha')
```

```python
def add(first, second):
    return first + second
print(add(4, 6))
```

```python
def square(number):
    return number ** 2
print(square(8))
```

```python
def welcome(name='Guest'):
    return f'Welcome, {name}'
print(welcome())
print(welcome('Ravi'))
```

```python
def rectangle_area(length, width):
    return length * width
print(rectangle_area(width=5, length=8))
```

```python
def is_even(number):
    return number % 2 == 0
print(is_even(12))
```

```python
def get_full_name(first, last):
    return f'{first.title()} {last.title()}'
print(get_full_name('mira', 'sen'))
```

```python
def repeat_text(text, times):
    return text * times
print(repeat_text('Go! ', 3))
```

```python
def calculate_total(price, tax_rate=0.18):
    return price + price * tax_rate
print(calculate_total(1000))
```
