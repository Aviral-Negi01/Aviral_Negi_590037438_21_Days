```python
text = 'programming'
print(text[0:6]) #slicing : text[start:end] → start is included, end is excluded.
print(text[6:])  #There is no ending index, so Python means start at index 6 and continue until the end of the string.


print(text[5:5]) # print noting
print(text[5:11])
```

```python
name = 'ravi'
print(name.capitalize()) # capitalize the first character
print(name.center(12, '-'))#Put "ravi" in the center of a string that is 12 characters wide, and use - to fill the remaining space.

name = name.capitalize()
print(name)

name = name.center(20,'~')
print(name)
```

```python
product = 'Laptop Pro'
price = 55000
slug = product.lower().replace(' ', '-') #The lower() method converts all letters to lowercase and Replace every space ' ' with a hyphen '-'.
print(f'{slug}: Rs. {price:,}') #The f before the string means this is an f-string.It allows us to put variables directly inside {}.
# in {price;,}:, adds commas to a number


name = 'Chicken Pizza'
price = 1000
bill = name.lower().replace(' ', '~')
print(f'{bill}: {price:,}')
```
